```python
# =============================================
# ⌨️  USER INPUT — MAKING PROGRAMS INTERACTIVE
# =============================================

# input() always returns a STRING
# You must convert it if you need math

print("🏏 CRICKET BATTING AVERAGE CALCULATOR")
print("=" * 40)

player_name   = input("Enter player name: ")
total_runs    = int(input("Enter total runs: "))
total_innings = int(input("Enter total innings: "))
not_outs      = int(input("Enter not outs: "))

# Calculate
actual_innings = total_innings - not_outs
batting_avg    = total_runs / actual_innings

print()
print(f"Player   : {player_name}")
print(f"Total Runs : {total_runs}")
print(f"Innings  : {actual_innings}")
print(f"Average  : {batting_avg:.2f}")

# Grade the average
if batting_avg >= 50:
    grade = "World Class 🌟"
elif batting_avg >= 40:
    grade = "Very Good ✅"
elif batting_avg >= 30:
    grade = "Average 👍"
else:
    grade = "Needs Work 💪"

print(f"Rating   : {grade}")
```

```python
print("🍕 ZOMATO DELIVERY BILL CALCULATOR")
print("=" * 40)

# Get details from user
restaurant = input("Enter restaurant name: ")
items = int(input("Enter number of items: "))
price = float(input("Enter price per item: ₹"))

order_type = input("Did you order veg or non-veg? ").lower()

# Calculations
subtotal = items * price
gst = subtotal * 0.18
delivery_charge = 40

# Non-veg surcharge example
if order_type == "non-veg":
    surcharge = 20
else:
    surcharge = 0

total = subtotal + gst + delivery_charge + surcharge

# Print receipt
print("\n")
print("=" * 35)
print("       ZOMATO ORDER RECEIPT")
print("=" * 35)

print(f"Restaurant : {restaurant}")
print(f"Items      : {items} × ₹{price:.0f} each")
print(f"Subtotal   : ₹{subtotal:.0f}")
print(f"GST (18%)  : ₹{gst:.0f}")
print(f"Delivery   : ₹{delivery_charge}")
print(f"Surcharge  : ₹{surcharge}")
print("-" * 35)
print(f"TOTAL      : ₹{total:.0f}")
print("=" * 35)
```
